```python
import pandas as pd

fact = pd.read_csv("data/gold/Fact_sales.csv")

print("FACT")
print("Rows:", len(fact))

print("\nUnique orders:", fact["order_id"].nunique())
print("Unique customers:", fact["customer_id"].nunique())

print("\nOrders per customer:")
print(
    fact.groupby("customer_id")["order_id"]
    .nunique()
    .describe()
)

print("\nCustomers with multiple orders:")
print(
    (fact.groupby("customer_id")["order_id"].nunique() > 1).sum()
)

print("\nOrder/customer relationship:")
print(
    fact.groupby("order_id")["customer_id"]
    .nunique()
    .describe()
)
```

```python
print(
    fact[
        ["order_id", "customer_id"]
    ]
    .drop_duplicates()
    .groupby("customer_id")
    .size()
    .value_counts()
    .sort_index()
)
```

```python
import pandas as pd

fact = pd.read_csv("data/gold/Fact_sales.csv")

print("FACT SHAPE:", fact.shape)

print("\nCOLUMNS:")
print(fact.columns.tolist())

print("\nUnique orders:")
print(fact["order_id"].nunique())

print("\nUnique customer IDs:")
print(fact["customer_id"].nunique())

print("\nUnique real customers:")
print(fact["customer_unique_id"].nunique())

print("\nCustomers with multiple orders:")
print(
    fact.groupby("customer_unique_id")["order_id"]
    .nunique()
    .gt(1)
    .sum()
)
```
