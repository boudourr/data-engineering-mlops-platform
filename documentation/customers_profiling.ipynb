```python
import pandas as pd
import numpy as np

from IPython.display import display
```

```python
file_path = "../data/raw/customers.csv"

customers = pd.read_csv(file_path)

print("Dataset loaded successfully.")
```

```python
print("="*60)
print("DATASET DIMENSIONS")
print("="*60)

rows, columns = customers.shape

print(f"Rows    : {rows}")
print(f"Columns : {columns}")
```

```python
print("="*60)
print("COLUMN NAMES")
print("="*60)

display(pd.DataFrame({
    "Column Name": customers.columns
}))
```

```python
print("="*60)
print("DATA TYPES")
print("="*60)

display(pd.DataFrame({
    "Column": customers.columns,
    "Data Type": customers.dtypes.values
}))
```

```python
print("="*60)
print("MEMORY USAGE")
print("="*60)

memory = customers.memory_usage(deep=True)

display(pd.DataFrame({
    "Column": memory.index,
    "Memory (Bytes)": memory.values
}))

print(f"\nTotal Memory : {memory.sum()/1024:.2f} KB")
```

```python
print("="*60)
print("MISSING VALUES")
print("="*60)

missing = customers.isnull().sum()

display(pd.DataFrame({
    "Column": missing.index,
    "Missing Values": missing.values
}))
```

```python
missing_percentage = (
    customers.isnull().mean()*100
).round(2)

display(pd.DataFrame({
    "Column": missing_percentage.index,
    "Missing %": missing_percentage.values
}))
```

```python
print("="*60)
print("DUPLICATE ROWS")
print("="*60)

duplicates = customers.duplicated().sum()

print("Duplicate Rows :", duplicates)
```

```python
primary_key = "customer_id"

duplicate_pk = customers[primary_key].duplicated().sum()

print("="*60)
print("PRIMARY KEY VALIDATION")
print("="*60)

print("Primary Key :", primary_key)
print("Duplicate Primary Keys :", duplicate_pk)
```

```python
unique_values = customers.nunique()

display(pd.DataFrame({
    "Column": unique_values.index,
    "Unique Values": unique_values.values
}))
```

```python
cardinality = (
    customers.nunique()/len(customers)
).round(3)

display(pd.DataFrame({
    "Column": cardinality.index,
    "Cardinality": cardinality.values
}))
```

```python
print("="*60)
print("STATISTICAL SUMMARY")
print("="*60)

display(customers.describe())
```

```python
for column in customers.select_dtypes(include="object").columns:

    print("="*60)
    print(column)
    print("="*60)

    display(
        customers[column]
        .value_counts()
        .head(10)
    )
```

```python
orders = pd.read_csv("../data/raw/orders.csv")

invalid_customers = (
    ~orders["customer_id"]
    .isin(customers["customer_id"])
).sum()

print("Orders without matching customer :", invalid_customers)
```

```python
total_cells = customers.shape[0] * customers.shape[1]

missing_cells = customers.isnull().sum().sum()

completeness = (
    (1 - missing_cells/total_cells)
    *100
)

duplicate_penalty = (
    customers.duplicated().sum()
)

quality = completeness

if duplicate_penalty > 0:
    quality -= 5

quality = max(0, round(quality,2))

print("="*60)
print("DATA QUALITY SCORE")
print("="*60)

print(f"Completeness : {completeness:.2f}%")
print(f"Overall Score : {quality:.2f}%")
```

```python
summary = pd.DataFrame({

    "Metric":[
        "Rows",
        "Columns",
        "Duplicate Rows",
        "Duplicate Primary Keys",
        "Missing Values",
        "Data Quality Score"
    ],

    "Value":[
        rows,
        columns,
        duplicates,
        duplicate_pk,
        missing.sum(),
        quality
    ]

})

display(summary)
```
