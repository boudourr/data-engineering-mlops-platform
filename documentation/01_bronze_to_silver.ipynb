**Validate primary key**

```python
products["product_id"].nunique() == len(products)
```

```python
products[
   (products["product_weight_g"] <= 0) |
   (products["product_length_cm"] <= 0) |
   (products["product_height_cm"] <= 0) |
   (products["product_width_cm"] <= 0)
]
```

**Convert count variables to intege**

```python
columns_to_convert = [
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty"
]

products[columns_to_convert] = products[columns_to_convert].astype("Int64")
```
