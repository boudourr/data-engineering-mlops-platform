```python
import pandas as pd
import numpy as np

from IPython.display import display
```

```python
file_path = "../data/raw/orders.csv"

orders = pd.read_csv(file_path)

print("Orders dataset loaded successfully.")
```

```python
print("="*60)
print("DATASET DIMENSIONS")
print("="*60)

rows, columns = orders.shape

print(f"Rows    : {rows}")
print(f"Columns : {columns}")
```

```python
display(pd.DataFrame({
    "Column Name": orders.columns
}))
```

```python
display(pd.DataFrame({
    "Column": orders.columns,
    "Data Type": orders.dtypes.values
}))
```

```python
memory = orders.memory_usage(deep=True)

display(pd.DataFrame({
    "Column": memory.index,
    "Memory (Bytes)": memory.values
}))

print(f"Total Memory : {memory.sum()/1024:.2f} KB")
```

```python
missing = orders.isnull().sum()

display(pd.DataFrame({
    "Column": missing.index,
    "Missing Values": missing.values
}))
```

```python
missing_percentage = (
    orders.isnull().mean()*100
).round(2)

display(pd.DataFrame({
    "Column": missing_percentage.index,
    "Missing %": missing_percentage.values
}))
```

```python
duplicates = orders.duplicated().sum()

print("Duplicate Rows :", duplicates)
```

```python
primary_key = "order_id"

duplicate_pk = orders[primary_key].duplicated().sum()

print("Duplicate Primary Keys :", duplicate_pk)
```

```python
unique_values = orders.nunique()

display(pd.DataFrame({
    "Column": unique_values.index,
    "Unique Values": unique_values.values
}))
```

```python
cardinality = (
    orders.nunique()/len(orders)
).round(3)

display(pd.DataFrame({
    "Column": cardinality.index,
    "Cardinality": cardinality.values
}))
```

```python
display(orders.describe(include="all"))
```

```python
for column in orders.select_dtypes(include=["object"]).columns:

    print("="*60)
    print(column)
    print("="*60)

    display(
        orders[column]
        .value_counts(dropna=False)
        .head(10)
    )
```

```python
customers = pd.read_csv("../data/raw/customers.csv")
```

```python
invalid_customer = (
    ~orders["customer_id"]
    .isin(customers["customer_id"])
).sum()

print("Orders without matching customer :", invalid_customer)
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")
```

```python
for col in date_columns:

    print("="*60)
    print(col)

    print("Minimum :", orders[col].min())
    print("Maximum :", orders[col].max())
```

```python
status = (
    orders["order_status"]
    .value_counts()
    .reset_index()
)

status.columns = [
    "Order Status",
    "Count"
]

display(status)
```

**Delivery Time Analysis**

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"] -
    orders["order_purchase_timestamp"]
).dt.days
```

```python
orders["delivery_days"].describe()
```

**Late Delivery Analysis**

```python
orders["late_delivery"] = (
    orders["order_delivered_customer_date"] >
    orders["order_estimated_delivery_date"]
)

orders["late_delivery"].value_counts()
```

**Datat Quality Score**

```python
total_cells = orders.shape[0] * orders.shape[1]

missing_cells = orders.isnull().sum().sum()

completeness = (
    (1 - missing_cells/total_cells)
    *100
)

duplicate_penalty = (
    orders.duplicated().sum()
)

quality = completeness

if duplicate_penalty > 0:
    quality -= 5

quality = max(0, round(quality,2))

print(f"Overall Data Quality : {quality}%")
```

**Profiling Summary**

```python
summary = pd.DataFrame({

    "Metric":[
        "Rows",
        "Columns",
        "Duplicate Rows",
        "Duplicate Primary Keys",
        "Missing Values",
        "Average Delivery Days",
        "Data Quality Score"
    ],

    "Value":[
        rows,
        columns,
        duplicates,
        duplicate_pk,
        missing.sum(),
        round(orders["delivery_days"].mean(),2),
        quality
    ]

})

display(summary)
```
